```python
from openapi_parser import parse
spec = parse("data/route.yaml")
spec.paths[0]
```

```python
responses = spec.paths[0].operations[0].responses
# res_ok = next((res for res in responses if res.code == 200), None)
res_ok = next(filter(lambda res: res.code == 200, responses))
res_ok.content[0].schema.type
```

```python
operation = spec.paths[0].operations[0]
responses = operation.responses
for response in responses:
    print(f"Status Code: {response.code}, Description: {response.description}")
```

```python
first_response = operation.responses[0]
print(f"Code: {first_response.code}")
print(f"Description: {first_response.description}")
print(f"Content: {first_response.content}")
```

```python
# Get the 200 response schema
res_ok_schema = res_ok.content[0].schema

# Navigate to the users array items
users_property = next(p for p in res_ok_schema.properties if p.name == "users")
user_items_schema = users_property.schema.items

# If items reference is a string (unlikely with this parser), resolve it
# Otherwise, the parser should have already resolved the $ref to the actual User schema
print(f"User schema type: {user_items_schema.type}")
print(f"User schema properties: {[p.name for p in user_items_schema.properties]}")

# Access actual User schema from spec
user_schema = spec.schemas.get("User")
print(f"\nResolved User schema properties:")
for prop in user_schema.properties:
    print(f"  - {prop.name}: {prop.schema.type}")
```
